# Task - 3A 

```python
def calculate_serial_sum(n:int):
    """
    Calculate the sum of numbers from 0-n
    Paramerter:
        n(int): number
    Returns:
        sum(int): 
    """
    sum  = 0
    for i in range(n+1):
        sum+=i

    return sum


def calculate_sum(m: int, n: int, res: list):
    sum=0
    for i in range(int(m+1),int(n+1)):
        sum+=i
    res.append(sum)
```

```python
import time
import threading

result = []

serial_start_time= time.time()
n= 100000000
serial_sum=calculate_serial_sum(n)
serial_total_time= time.time()-serial_start_time

print(f"Sum of first {n} is: {serial_sum}")
print("Time Taken:", serial_total_time,"seconds")
```

# Task-3B

```python
num_thread=4
threads=[]
x=0
mult=n/num_thread
threat_st = time.time()
for i in range(0,num_thread):
    x=mult*i
    y= mult*(i+1)
    thread = threading.Thread(target=calculate_sum, args=(x,y,result))
    threads.append(thread)

for thread in threads:
    thread.start()

for thread in threads:
    thread.join()

thread_sum = sum(result)
thread_total_time= time.time()-threat_st



print("Parallel Results")
print(f"Sum of first {n} is: {thread_sum}")
print("Time Taken:", thread_total_time,"seconds")


thread_speedup = (serial_total_time)/(thread_total_time)


print(f"Speedup: {thread_speedup}")
print(f"Efficency: {thread_speedup/num_thread}")
```

# Task-3C

```python
def calculate_process_sum(m: int, n: int, queue):
    sum=0
    for i in range(int(m+1),int(n+1)):
        sum+=i
    queue.put(sum)
```

```python
import multiprocessing

num_process= 4

mult= n // num_thread

processes = []
q = multiprocessing.Queue()

for i in range(num_process):
    x=mult*i
    y= mult*(i+1)
    process = multiprocessing.Process(target=calculate_process_sum,args=(x,y,q))
    processes.append(process)


process_start = time.time()

for process in processes:
    process.start()

for process in processes:
    process.join()

process_sum =0

for i in range(num_process):
    process_sum+=q.get()
process_total_time= time.time()-process_start



print("Process Results")
print(f"Sum of first {n} is: {process_sum}")
print("Time Taken:", process_total_time,"seconds")


process_speedup = serial_total_time/process_total_time


print(f"Speedup: {process_speedup}")
print(f"Efficency: {process_speedup/num_process}")
```
