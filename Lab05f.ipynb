```python
import time
import multiprocessing
from concurrent.futures import ProcessPoolExecutor
import random

# Function to compute the square of a number
def square(n):
    return n * n

# List of random numbers
numbers_103 = [random.randint(1, 100) for _ in range(103)]
numbers_107 = [random.randint(1, 100) for _ in range(10**7)]
```

```python
def sequential_square(numbers):
    return [square(n) for n in numbers]

start = time.time()
sequential_square(numbers_103)
print("Sequential for loop (103 numbers):", time.time() - start)
```

```python
def multiprocessing_square(numbers):
    processes = []
    results = multiprocessing.Manager().list()  # Using Manager for shared results

    for number in numbers:
        process = multiprocessing.Process(target=lambda n, res: res.append(square(n)), args=(number, results))
        processes.append(process)
        process.start()

    for process in processes:
        process.join()

    return list(results)

start = time.time()
multiprocessing_square(numbers_103)
print("Multiprocessing for loop (103 numbers):", time.time() - start)
```

```python
def pool_map_square(numbers):
    pool_size = multiprocessing.cpu_count()  # Optimal number of workers = CPU cores
    with multiprocessing.Pool(pool_size) as pool:
        return pool.map(square, numbers)

start = time.time()
pool_map_square(numbers_103)
print("Multiprocessing pool map (103 numbers):", time.time() - start)
```

```python
def pool_apply_square(numbers):
    with multiprocessing.Pool() as pool:
        results = [pool.apply_async(square, (n,)) for n in numbers]
        return [result.get() for result in results]

start = time.time()
pool_apply_square(numbers_103)
print("Multiprocessing pool apply_async (103 numbers):", time.time() - start)
```

```python
def process_pool_square(numbers):
    with ProcessPoolExecutor() as executor:
        return list(executor.map(square, numbers))

start = time.time()
process_pool_square(numbers_103)
print("ProcessPoolExecutor (103 numbers):", time.time() - start)
```

```python
# For sequential
start = time.time()
sequential_square(numbers_107)
print("Sequential for loop (10⁷ numbers):", time.time() - start)

# For multiprocessing loop
start = time.time()
multiprocessing_square(numbers_107)
print("Multiprocessing for loop (10⁷ numbers):", time.time() - start)

# For pool with map
start = time.time()
pool_map_square(numbers_107)
print("Multiprocessing pool map (10⁷ numbers):", time.time() - start)

# For pool with apply_async
start = time.time()
pool_apply_square(numbers_107)
print("Multiprocessing pool apply_async (10⁷ numbers):", time.time() - start)

# For ProcessPoolExecutor
start = time.time()
process_pool_square(numbers_107)
print("ProcessPoolExecutor (10⁷ numbers):", time.time() - start)
```
